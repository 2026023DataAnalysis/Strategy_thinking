## Physical constants and the mass of one N₂ molecule

```python
K_B = constants.k             # Boltzmann constant, J/K
N_A = constants.N_A           # Avogadro constant, 1/mol

M_N2 = 2 * 14.0067e-3         # molar mass of N2, kg/mol
M_MOLECULE = M_N2 / N_A       # mass of ONE N2 molecule, kg
```

- 2 * 14.0067e-3: N₂ has two nitrogen atoms of 14.0067 g/mol each. e-3 converts grams to kilograms, which is the SI unit.
- M_N2 / N_A: one mole holds 6.022 × 10²³ molecules, so dividing the mass of a mole by that number gives the mass of one molecule.
- Capital letters (K_B, M_MOLECULE): a Python convention meaning 'this is a constant, never change it'. 

## The Maxwell-Boltzmann speed distribution

$$f(v) = 4\pi \left(\frac{m}{2\pi k_BT}\right)^{3/2} v^2\, \exp\!\left(-\frac{m v^2}{2k_BT}\right)$$

$f(v)$ is a **probability density**: the probability that a molecule has a speed between $v_1$ and $v_2$ is the area under the curve, $\int_{v_1}^{v_2} f(v)\,dv$. So the total area from 0 to infinity must be exactly 1.

```python
from scipy import integrate

def maxwell_pdf(v, T, m=M_MOLECULE):
    a = m / (2 * K_B * T)
    return 4 * np.pi * (a / np.pi) ** 1.5 * v**2 * np.exp(-a * v**2)

for T in TEMPERATURES:
    area = integrate.quad(maxwell_pdf, 0, np.inf, args=(T,))[0]
    print(f"T = {T:>4} K   total area under f(v) = {area:.10f}")
```

- `def maxwell_pdf(v, T, m=M_MOLECULE)`: the formula above as a Python function. `m=M_MOLECULE` is a default value, so we only need to give the speed and the temperature.
- `a = m / (2 * K_B * T)`: a short name for $\frac{m}{2k_BT}$, which appears twice in the formula. Then $f(v) = 4\pi\left(\frac{a}{\pi}\right)^{3/2} v^2 e^{-a v^2}$.
- `v**2`: the **geometric factor**. Many velocity directions have the same speed $v$; they lie on a sphere of area $4\pi v^2$. This term pushes the probability **up** at higher speeds.
- `np.exp(-a * v**2)`: the **Boltzmann factor** $e^{-E_k/k_BT}$ with $E_k = \frac12 mv^2$. High speed means high energy, and high energy is less probable. This term pushes the probability **down** at higher speeds.
- The competition between the two terms creates one peak: the **most probable speed**.
- `integrate.quad(maxwell_pdf, 0, np.inf, args=(T,))`: numerical integration from 0 to infinity. `args=(T,)` passes the temperature to the function. `[0]` keeps the value and drops the error estimate.
- **Result:** the area is 1.0000000000 at all four temperatures, so the formula is a correct probability distribution. This also checks that the constants and the mass in kg are consistent.

```python
def v_most_probable(T, m=M_MOLECULE):
    return np.sqrt(2 * K_B * T / m)

def v_mean_theory(T, m=M_MOLECULE):
    return np.sqrt(8 * K_B * T / (np.pi * m))

def v_rms_theory(T, m=M_MOLECULE):
    return np.sqrt(3 * K_B * T / m)

theory = pd.DataFrame({
    "T (K)": TEMPERATURES,
    "v_mp (m/s)": v_most_probable(np.array(TEMPERATURES)),
    "v_mean (m/s)": v_mean_theory(np.array(TEMPERATURES)),
    "v_rms (m/s)": v_rms_theory(np.array(TEMPERATURES)),
}).set_index("T (K)")

theory.round(1)
```

- Each function is one formula from the heading. They work with a single temperature or with a whole NumPy array of temperatures.
- `np.array(TEMPERATURES)`: converts the list to an array, so each function calculates all four temperatures in one step.
- `pd.DataFrame({...}).set_index("T (K)")`: puts the results in a table with one row per temperature.
- **Result 1, the order is always $v_{mp} < v_{mean} < v_{rms}$.** At 300 K: 422.0 < 476.2 < 516.8 m/s. The mean speed of N₂ at room temperature is about 476 m/s, about 1,700 km/h.
- **Why this order?** $v_{mp} < v_{mean}$ because the distribution is **right-skewed**: the long tail of fast molecules pulls the mean to the right of the peak. $v_{mean} < v_{rms}$ for any spread-out distribution, because $\langle v^2\rangle - \langle v\rangle^2 = \sigma^2 > 0$.
- **The ratio never changes:** $v_{mp} : v_{mean} : v_{rms} = 1 : 1.128 : 1.225$ at every temperature, because all three are a constant times $\sqrt{T/m}$.
- **Result 2, speed grows like $\sqrt{T}$, not like $T$.** From 100 K to 1000 K the temperature is ×10, but the mean speed only goes from 274.9 to 869.4 m/s, ×3.16 $= \sqrt{10}$. So the relationship between speed and temperature is **not linear**. This is important for the Linear Regression later.

## SciPy maxwell parameters

SciPy defines the standard Maxwell distribution without units: $f(x) = \sqrt{\frac{2}{\pi}}\,x^2 e^{-x^2/2}$. With `loc = 0` and `scale = a`, it uses $x = v/a$, so

$$f(v) = \sqrt{\frac{2}{\pi}}\,\frac{v^2}{a^3}\,e^{-v^2/2a^2}$$

This is our physics formula when the exponents are equal: $\frac{v^2}{2a^2} = \frac{mv^2}{2k_BT}$, so

$$a = \sqrt{\frac{k_BT}{m}}$$

Before we generate any data, we check that SciPy with this `scale` gives exactly our formulas.

```python
def maxwell_scale(T, m=M_MOLECULE):
    return np.sqrt(K_B * T / m)

v_grid = np.linspace(0, 3000, 2001)
rows = []
for T in TEMPERATURES:
    dist = stats.maxwell(loc=0, scale=maxwell_scale(T))
    rows.append({
        "T (K)": T,
        "scale a (m/s)": maxwell_scale(T),
        "SciPy mean (m/s)": dist.mean(),
        "formula v_mean (m/s)": v_mean_theory(T),
        "max PDF difference": np.max(np.abs(dist.pdf(v_grid) - maxwell_pdf(v_grid, T))),
    })

pd.DataFrame(rows).set_index("T (K)")
```

- `maxwell_scale(T)`: the SciPy `scale` parameter $a = \sqrt{k_BT/m}$. Physically, $a$ is the standard deviation of **one velocity component** ($v_x$, $v_y$ or $v_z$).
- `loc=0`: speeds start at 0. A different `loc` would shift the whole curve, which has no physical meaning.
- `stats.maxwell(loc=0, scale=...)`: a "frozen" distribution with fixed parameters. `.mean()` and `.pdf(v)` give its mean and its density.
- `max PDF difference`: the largest gap between SciPy's curve and our `maxwell_pdf` over 2,001 speeds.

- **Conclusion:** `stats.maxwell(loc=0, scale=√(k_B T/m))` **is** the Maxwell-Boltzmann distribution of N₂. Now we can safely use SciPy to generate the data.
- If we had passed `scale=T` by mistake, SciPy would still produce numbers, but they would be physically wrong. This check protects us against that silent error.

```python
heights = stats.maxwell(loc=0, scale=  maxwell_scale(T)) 
heights.mean()
```

## Generating the molecular speeds

For each temperature we draw 10,000 random speeds from `stats.maxwell(loc=0, scale=a)` with $a = \sqrt{k_BT/m}$. Each row of the new table is **one N₂ molecule**: its temperature (the class label) and its speed.

The NIST table stays in `df`. The simulated molecules go in a new table called `mb` (Maxwell-Boltzmann).

```python
N_PER_T = 10_000                          # molecules per temperature
rng_data = np.random.default_rng(SEED)    # fresh generator: same data every time this cell runs

frames = []
for T in TEMPERATURES:
    speeds = stats.maxwell.rvs(loc=0, scale=maxwell_scale(T), size=N_PER_T, random_state=rng_data)
    frames.append(pd.DataFrame({"Temperature_K": T, "Speed": speeds}))

mb = pd.concat(frames, ignore_index=True)

print("shape:", mb.shape)
print(mb.head(3), "\n")
mb.groupby("Temperature_K")["Speed"].agg(["count", "mean", "min", "max"]).round(1)
```

- `N_PER_T = 10_000`: 10,000 molecules per temperature. The `_` is only for reading; Python sees `10000`.
- `rng_data = np.random.default_rng(SEED)`: a new generator created **inside this cell**. Even if other cells use random numbers before it, this cell always produces the same 40,000 speeds.
- `stats.maxwell.rvs(...)`: `rvs` = "random variates", i.e. random samples from the distribution. `random_state=rng_data` makes SciPy use our seeded generator.
- `pd.DataFrame({"Temperature_K": T, "Speed": speeds})`: one small table per temperature. The single value `T` is repeated for all 10,000 rows.
- `pd.concat(..., ignore_index=True)`: stacks the four tables into one and numbers the rows 0 to 39,999.
- **Result:** `mb` has 40,000 rows and 2 columns, with exactly 10,000 molecules per temperature (balanced classes).
- The sample means (275.9, 476.3, 610.1 and 870.3 m/s) are close to the theoretical $v_{mean}$ (274.9, 476.2, 614.7 and 869.4 m/s). The small differences are random sampling error. The next cell checks this properly.
- The minimum speeds are close to 0 and the maximum speeds are about 3 times $v_{mean}$. This is the long right tail we saw in the curves.

## Checking the sample against theory

A random sample never matches theory exactly. To check whether a gap is just normal random error, we compare it with the **standard error** of the mean:

$$SE = \frac{s}{\sqrt{n}}, \qquad z = \frac{\bar{v} - v_{mean}}{SE}$$

If the sample really comes from the theoretical distribution, $|z|$ is usually below 2 (about 95% of the time).

```python
rows = []
for T, g in mb.groupby("Temperature_K"):
    v = g["Speed"]
    se = v.std() / np.sqrt(len(v))
    v_rms_sample = np.sqrt(np.mean(v**2))
    rows.append({
        "T (K)": T,
        "sample mean": v.mean(),
        "theory v_mean": v_mean_theory(T),
        "SE": se,
        "z": (v.mean() - v_mean_theory(T)) / se,
        "sample v_rms": v_rms_sample,
        "theory v_rms": v_rms_theory(T),
        "v_rms error (%)": 100 * (v_rms_sample / v_rms_theory(T) - 1),
    })

pd.DataFrame(rows).set_index("T (K)").round(2)
```

- `mb.groupby("Temperature_K")`: loops over the four temperatures. `T` is the temperature and `g` holds its 10,000 rows.
- `se = v.std() / np.sqrt(len(v))`: the standard error. It is the typical size of the random error in a mean of 10,000 values.
- `z`: how many standard errors the sample mean is away from the theoretical mean.
- `np.sqrt(np.mean(v**2))`: the sample root-mean-square speed. It is the same definition as $v_{rms}$, applied to data.
- **Result:** all four $|z|$ values are below 2 (0.87, 0.04, -1.80 and 0.26). The gaps between the sample and theory are the size we expect from random sampling.
- 500 K has the largest gap (-4.6 m/s, $z = -1.80$). It is still inside the ±2 range, so it is not a problem. This seed simply gave slightly slow molecules at 500 K.
- The sample $v_{rms}$ is within 1% of theory at every temperature (largest error -0.80% at 500 K).
- The SE grows with temperature (1.16 to 3.71 m/s), because hotter gases have more spread. In relative terms it is always about 0.42%, the same CV for every temperature.
- **Conclusion:** the simulated data agrees with the Maxwell-Boltzmann theory. We can use `mb` for the analysis.

## Physical columns: speed squared and kinetic energy

The kinetic energy of one molecule is

$$E_k = \tfrac{1}{2} m v^2$$

The equipartition theorem says the **average** kinetic energy depends only on the temperature, not on the type of gas:

$$\langle E_k \rangle = \tfrac{3}{2} k_B T$$

The $3$ comes from the three directions of motion ($x$, $y$, $z$), and each direction carries $\tfrac{1}{2}k_BT$ on average.

```python
mb
```

```python
mb["Speed_Squared"] = mb["Speed"] ** 2                          # m^2/s^2
mb["Kinetic_Energy"] = 0.5 * M_MOLECULE * mb["Speed_Squared"]   # J

energy = mb.groupby("Temperature_K")["Kinetic_Energy"].mean().to_frame("mean E_k (J)")
energy["3/2 k_B T (J)"] = 1.5 * K_B * energy.index
energy["mean E_k / (k_B T)"] = energy["mean E_k (J)"] / (K_B * energy.index)

print(mb.head(3), "\n")
energy.style.format({"mean E_k (J)": "{:.4e}", "3/2 k_B T (J)": "{:.4e}", "mean E_k / (k_B T)": "{:.4f}"})
```

- `mb["Speed_Squared"]`: $v^2$ for every molecule, in m²/s².
- `mb["Kinetic_Energy"]`: $\tfrac{1}{2} m v^2$ in joules. The values are tiny (about $10^{-21}$ J) because one molecule is very light.
- `energy.index`: the four temperatures, so `1.5 * K_B * energy.index` gives $\tfrac{3}{2}k_BT$ for each row.
- `mean E_k / (k_B T)`: the energy measured in units of $k_BT$. Theory says it should be **1.5** at every temperature.
- **Result:** the ratios are 1.5108, 1.5076, 1.4760 and 1.5076. All of them are close to 1.5, so equipartition holds in our data.
- The energy errors (+0.72%, +0.51%, -1.60%, +0.51%) are exactly **twice** the $v_{rms}$ errors from the previous cell, because $\langle E_k \rangle = \tfrac{1}{2} m\, v_{rms}^2$. A squared quantity doubles the relative error.
- 500 K is again the furthest from theory, for the same reason as before (this sample has slightly slow molecules).
- **Note for later:** `Speed_Squared` and `Kinetic_Energy` are calculated only from `Speed`. They add **no new information**, so we will not use all three as separate model features.

## Theory columns, normalised speed and data leakage

We add the theoretical speeds for each molecule's temperature, plus the **normalised speed**:

$$u = \frac{v}{v_{mp}(T)}$$

These columns are useful for checking the physics. But they are calculated **from the temperature itself**, so a model that sees them would already know the answer. This is called **data leakage**. They must **never** be model inputs.

```python
mb["V_mp_theory"] = v_most_probable(mb["Temperature_K"])
mb["V_mean_theory"] = v_mean_theory(mb["Temperature_K"])
mb["V_rms_theory"] = v_rms_theory(mb["Temperature_K"])
mb["Normalised_Speed"] = mb["Speed"] / mb["V_mp_theory"]

LEAKAGE_COLS = ["V_mp_theory", "V_mean_theory", "V_rms_theory", "Normalised_Speed"]   # never use as model inputs

summary = mb.groupby("Temperature_K").agg(
    V_mp_unique=("V_mp_theory", "nunique"),
    V_mp_theory=("V_mp_theory", "first"),
    Norm_mean=("Normalised_Speed", "mean"),
    Norm_std=("Normalised_Speed", "std"),
)
print("columns:", list(mb.columns))
print(f"theory for u: mean = {2 / np.sqrt(np.pi):.4f}, std = {np.sqrt(1.5 - 4 / np.pi):.4f}\n")
summary.round(4)
```

- `v_most_probable(mb["Temperature_K"])`: our functions also work on a whole column, so every row gets the theory value for its own temperature.
- `Normalised_Speed`: each speed divided by $v_{mp}$ of its temperature. $u = 1$ means "exactly at the peak".
- `LEAKAGE_COLS`: a list we will use later to **remove** these columns before any model is trained.
- `.agg(...)`: several summaries at once. `nunique` counts different values, and `first` takes the first value.
- **Result 1 (leakage):** `V_mp_unique = 1` for every temperature. Each theory column has exactly **one value per class**, so it works as a perfect lookup table for the temperature. A model would score 100% accuracy and learn nothing about the physics.
- **Result 2 (universal curve):** the mean of $u$ is about 1.13 and its standard deviation about 0.48 at **all four** temperatures. Theory gives 1.1284 and 0.4762. After normalising, every temperature has the **same distribution**.
- `Normalised_Speed` is a hidden leak: it looks like a measurement, but it needs $T$ to be calculated. Dividing by $v_{mp}(T)$ removes exactly the temperature information we want the model to find.
- `Temperature_K` is the **target** (the label). `Speed` is the real measured feature.

## Outliers with the IQR rule

The IQR rule marks a value as an outlier if it lies outside the **fences**:

$$IQR = Q_3 - Q_1, \qquad \text{lower fence} = Q_1 - 1.5\,IQR, \qquad \text{upper fence} = Q_3 + 1.5\,IQR$$

We apply it **separately for each temperature**, because a speed that is extreme at 100 K is normal at 1000 K. Then we compare the outlier percentage with what the Maxwell-Boltzmann theory predicts for its own tail.

```python
rows = []
for T, g in mb.groupby("Temperature_K"):
    v = g["Speed"]
    q1, q3 = v.quantile([0.25, 0.75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    dist = stats.maxwell(loc=0, scale=maxwell_scale(T))
    rows.append({
        "T (K)": T,
        "lower fence": low,
        "upper fence": high,
        "below lower": (v < low).sum(),
        "above upper": (v > high).sum(),
        "outliers (%)": 100 * ((v < low) | (v > high)).mean(),
        "theory above upper (%)": 100 * dist.sf(high),
    })

pd.DataFrame(rows).set_index("T (K)").round(2)
```

- `v.quantile([0.25, 0.75])`: the first quartile $Q_1$ and the third quartile $Q_3$ of the speeds at one temperature.
- `low, high`: the IQR fences, calculated separately for each temperature.
- `(v < low) | (v > high)`: `True` for an outlier. `|` means "or". `.mean()` of True/False values gives the fraction of outliers.
- `dist.sf(high)`: the **survival function**, $P(v > \text{upper fence}) = 1 - \text{cdf}$. It is the fraction of molecules the theory **expects** above the fence.
- **Result 1:** the lower fence is **negative** at every temperature, so no molecule can be below it. All outliers are on the fast side, because of the long right tail.
- **Result 2:** about 0.8% to 1.1% of the molecules are above the upper fence (76 to 107 per temperature). The theory predicts 0.8% to 1.0%, so the outliers are **expected by the physics**.
- **Decision: keep all outliers.** They are not errors. They are real fast molecules, and they are exactly the tail that makes Maxwell-Boltzmann asymmetric. Removing them would make the mean, $v_{rms}$ and the kinetic energy too low, and the data would no longer follow the theory.
- The IQR rule assumes a roughly **symmetric** distribution. For a skewed distribution it always flags part of the normal tail, so it is a check here, not a cleaning rule.

## Target, class encoding and feature lists

Before any model, we decide clearly **which column is the answer** and **which columns the model may see**:

- **Target:** `Temperature_K`, encoded as a class number (100 K → 0, 300 K → 1, 500 K → 2, 1000 K → 3).
- **Allowed feature (Level A, one molecule):** `Speed`.
- **Redundant:** `Speed_Squared` and `Kinetic_Energy`. They are calculated only from `Speed`.
- **Forbidden (leakage):** the four `LEAKAGE_COLS`.

```python
TARGET = "Temperature_K"
CLASS_LABELS = {T: i for i, T in enumerate(TEMPERATURES)}   # {100: 0, 300: 1, 500: 2, 1000: 3}
mb["Class"] = mb[TARGET].map(CLASS_LABELS)

REDUNDANT_COLS = ["Speed_Squared", "Kinetic_Energy"]   # calculated only from Speed
FEATURES_A = ["Speed"]                                 # Level A: one molecule
NEVER_INPUTS = LEAKAGE_COLS + [TARGET, "Class"]

assert not set(FEATURES_A) & set(NEVER_INPUTS), "leakage: a forbidden column is in the features"

print("class encoding:", CLASS_LABELS)
print("features (Level A):", FEATURES_A)
print("never model inputs:", NEVER_INPUTS)
print("redundant:", REDUNDANT_COLS, "\n")

print("Spearman correlation:")
print(mb[["Speed", "Speed_Squared", "Kinetic_Energy"]].corr(method="spearman").round(4), "\n")
mb["Class"].value_counts().sort_index().to_frame("count")
```

- `{T: i for i, T in enumerate(TEMPERATURES)}`: a dictionary comprehension. `enumerate` gives each temperature a position number, so 100 K → 0, 300 K → 1, 500 K → 2 and 1000 K → 3.
- `mb[TARGET].map(CLASS_LABELS)`: replaces each temperature with its class number in a new column `Class`.
- A simple number code is fine for the **target**. Classifiers treat 0, 1, 2 and 3 as names, not as amounts. (For an input feature with categories, we would need one-hot encoding instead.)
- `NEVER_INPUTS`: the leakage columns plus the target itself, in both forms (`Temperature_K` and `Class`).
- `assert ...`: a safety check. `set(A) & set(B)` is the overlap of the two lists. If a forbidden column were ever added to the features, the notebook would **stop with an error** here instead of silently giving a fake 100% accuracy.
- **Spearman correlation = 1.0** between `Speed`, `Speed_Squared` and `Kinetic_Energy`. Spearman compares the **ranking** of the values: the fastest molecule also has the largest $v^2$ and the largest $E_k$. The three columns hold the same information, so we keep only `Speed`.
- **Result:** 10,000 molecules in each class, so the classes are perfectly balanced. Accuracy is therefore a fair metric, and a model that guesses randomly would score 25%.
