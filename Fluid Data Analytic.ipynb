## Physical constants and the mass of one N₂ molecule

```python
K_B = constants.k             # Boltzmann constant, J/K
N_A = constants.N_A           # Avogadro constant, 1/mol

M_N2 = 2 * 14.0067e-3         # molar mass of N2, kg/mol
M_MOLECULE = M_N2 / N_A       # mass of ONE N2 molecule, kg
```

- 2 * 14.0067e-3: N₂ has two nitrogen atoms of 14.0067 g/mol each. e-3 converts grams to kilograms, which is the SI unit.
- M_N2 / N_A: one mole holds 6.022 × 10²³ molecules, so dividing the mass of a mole by that number gives the mass of one molecule.
- Capital letters (K_B, M_MOLECULE): a Python convention meaning 'this is a constant, never change it'. 

## The Maxwell-Boltzmann speed distribution

$$f(v) = 4\pi \left(\frac{m}{2\pi k_BT}\right)^{3/2} v^2\, \exp\!\left(-\frac{m v^2}{2k_BT}\right)$$

$f(v)$ is a **probability density**: the probability that a molecule has a speed between $v_1$ and $v_2$ is the area under the curve, $\int_{v_1}^{v_2} f(v)\,dv$. So the total area from 0 to infinity must be exactly 1.

```python
from scipy import integrate

def maxwell_pdf(v, T, m=M_MOLECULE):
    a = m / (2 * K_B * T)
    return 4 * np.pi * (a / np.pi) ** 1.5 * v**2 * np.exp(-a * v**2)

for T in TEMPERATURES:
    area = integrate.quad(maxwell_pdf, 0, np.inf, args=(T,))[0]
    print(f"T = {T:>4} K   total area under f(v) = {area:.10f}")
```

- `def maxwell_pdf(v, T, m=M_MOLECULE)`: the formula above as a Python function. `m=M_MOLECULE` is a default value, so we only need to give the speed and the temperature.
- `a = m / (2 * K_B * T)`: a short name for $\frac{m}{2k_BT}$, which appears twice in the formula. Then $f(v) = 4\pi\left(\frac{a}{\pi}\right)^{3/2} v^2 e^{-a v^2}$.
- `v**2`: the **geometric factor**. Many velocity directions have the same speed $v$; they lie on a sphere of area $4\pi v^2$. This term pushes the probability **up** at higher speeds.
- `np.exp(-a * v**2)`: the **Boltzmann factor** $e^{-E_k/k_BT}$ with $E_k = \frac12 mv^2$. High speed means high energy, and high energy is less probable. This term pushes the probability **down** at higher speeds.
- The competition between the two terms creates one peak: the **most probable speed**.
- `integrate.quad(maxwell_pdf, 0, np.inf, args=(T,))`: numerical integration from 0 to infinity. `args=(T,)` passes the temperature to the function. `[0]` keeps the value and drops the error estimate.
- **Result:** the area is 1.0000000000 at all four temperatures, so the formula is a correct probability distribution. This also checks that the constants and the mass in kg are consistent.

```python
def v_most_probable(T, m=M_MOLECULE):
    return np.sqrt(2 * K_B * T / m)

def v_mean_theory(T, m=M_MOLECULE):
    return np.sqrt(8 * K_B * T / (np.pi * m))

def v_rms_theory(T, m=M_MOLECULE):
    return np.sqrt(3 * K_B * T / m)

theory = pd.DataFrame({
    "T (K)": TEMPERATURES,
    "v_mp (m/s)": v_most_probable(np.array(TEMPERATURES)),
    "v_mean (m/s)": v_mean_theory(np.array(TEMPERATURES)),
    "v_rms (m/s)": v_rms_theory(np.array(TEMPERATURES)),
}).set_index("T (K)")

theory.round(1)
```

- Each function is one formula from the heading. They work with a single temperature or with a whole NumPy array of temperatures.
- `np.array(TEMPERATURES)`: converts the list to an array, so each function calculates all four temperatures in one step.
- `pd.DataFrame({...}).set_index("T (K)")`: puts the results in a table with one row per temperature.
- **Result 1, the order is always $v_{mp} < v_{mean} < v_{rms}$.** At 300 K: 422.0 < 476.2 < 516.8 m/s. The mean speed of N₂ at room temperature is about 476 m/s, about 1,700 km/h.
- **Why this order?** $v_{mp} < v_{mean}$ because the distribution is **right-skewed**: the long tail of fast molecules pulls the mean to the right of the peak. $v_{mean} < v_{rms}$ for any spread-out distribution, because $\langle v^2\rangle - \langle v\rangle^2 = \sigma^2 > 0$.
- **The ratio never changes:** $v_{mp} : v_{mean} : v_{rms} = 1 : 1.128 : 1.225$ at every temperature, because all three are a constant times $\sqrt{T/m}$.
- **Result 2, speed grows like $\sqrt{T}$, not like $T$.** From 100 K to 1000 K the temperature is ×10, but the mean speed only goes from 274.9 to 869.4 m/s, ×3.16 $= \sqrt{10}$. So the relationship between speed and temperature is **not linear**. This is important for the Linear Regression later.

## SciPy maxwell parameters

SciPy defines the standard Maxwell distribution without units: $f(x) = \sqrt{\frac{2}{\pi}}\,x^2 e^{-x^2/2}$. With `loc = 0` and `scale = a`, it uses $x = v/a$, so

$$f(v) = \sqrt{\frac{2}{\pi}}\,\frac{v^2}{a^3}\,e^{-v^2/2a^2}$$

This is our physics formula when the exponents are equal: $\frac{v^2}{2a^2} = \frac{mv^2}{2k_BT}$, so

$$a = \sqrt{\frac{k_BT}{m}}$$

Before we generate any data, we check that SciPy with this `scale` gives exactly our formulas.

```python
def maxwell_scale(T, m=M_MOLECULE):
    return np.sqrt(K_B * T / m)

v_grid = np.linspace(0, 3000, 2001)
rows = []
for T in TEMPERATURES:
    dist = stats.maxwell(loc=0, scale=maxwell_scale(T))
    rows.append({
        "T (K)": T,
        "scale a (m/s)": maxwell_scale(T),
        "SciPy mean (m/s)": dist.mean(),
        "formula v_mean (m/s)": v_mean_theory(T),
        "max PDF difference": np.max(np.abs(dist.pdf(v_grid) - maxwell_pdf(v_grid, T))),
    })

pd.DataFrame(rows).set_index("T (K)")
```

- `maxwell_scale(T)`: the SciPy `scale` parameter $a = \sqrt{k_BT/m}$. Physically, $a$ is the standard deviation of **one velocity component** ($v_x$, $v_y$ or $v_z$).
- `loc=0`: speeds start at 0. A different `loc` would shift the whole curve, which has no physical meaning.
- `stats.maxwell(loc=0, scale=...)`: a "frozen" distribution with fixed parameters. `.mean()` and `.pdf(v)` give its mean and its density.
- `max PDF difference`: the largest gap between SciPy's curve and our `maxwell_pdf` over 2,001 speeds.

- **Conclusion:** `stats.maxwell(loc=0, scale=√(k_B T/m))` **is** the Maxwell-Boltzmann distribution of N₂. Now we can safely use SciPy to generate the data.
- If we had passed `scale=T` by mistake, SciPy would still produce numbers, but they would be physically wrong. This check protects us against that silent error.

## Generating the molecular speeds

For each temperature we draw 10,000 random speeds from `stats.maxwell(loc=0, scale=a)` with $a = \sqrt{k_BT/m}$. Each row of the new table is **one N₂ molecule**: its temperature (the class label) and its speed.

The NIST table stays in `df`. The simulated molecules go in a new table called `mb` (Maxwell-Boltzmann).

```python
N_PER_T = 10_000                          # molecules per temperature
rng_data = np.random.default_rng(SEED)    # fresh generator: same data every time this cell runs

frames = []
for T in TEMPERATURES:
    speeds = stats.maxwell.rvs(loc=0, scale=maxwell_scale(T), size=N_PER_T, random_state=rng_data)
    frames.append(pd.DataFrame({"Temperature_K": T, "Speed": speeds}))

mb = pd.concat(frames, ignore_index=True)

print("shape:", mb.shape)
print(mb.head(3), "\n")
mb.groupby("Temperature_K")["Speed"].agg(["count", "mean", "min", "max"]).round(1)
```

- `N_PER_T = 10_000`: 10,000 molecules per temperature. The `_` is only for reading; Python sees `10000`.
- `rng_data = np.random.default_rng(SEED)`: a new generator created **inside this cell**. Even if other cells use random numbers before it, this cell always produces the same 40,000 speeds.
- `stats.maxwell.rvs(...)`: `rvs` = "random variates", i.e. random samples from the distribution. `random_state=rng_data` makes SciPy use our seeded generator.
- `pd.DataFrame({"Temperature_K": T, "Speed": speeds})`: one small table per temperature. The single value `T` is repeated for all 10,000 rows.
- `pd.concat(..., ignore_index=True)`: stacks the four tables into one and numbers the rows 0 to 39,999.
- **Result:** `mb` has 40,000 rows and 2 columns, with exactly 10,000 molecules per temperature (balanced classes).
- The sample means (275.9, 476.3, 610.1 and 870.3 m/s) are close to the theoretical $v_{mean}$ (274.9, 476.2, 614.7 and 869.4 m/s). The small differences are random sampling error. The next cell checks this properly.
- The minimum speeds are close to 0 and the maximum speeds are about 3 times $v_{mean}$. This is the long right tail we saw in the curves.

## Checking the sample against theory

A random sample never matches theory exactly. To check whether a gap is just normal random error, we compare it with the **standard error** of the mean:

$$SE = \frac{s}{\sqrt{n}}, \qquad z = \frac{\bar{v} - v_{mean}}{SE}$$

If the sample really comes from the theoretical distribution, $|z|$ is usually below 2 (about 95% of the time).

```python
rows = []
for T, g in mb.groupby("Temperature_K"):
    v = g["Speed"]
    se = v.std() / np.sqrt(len(v))
    v_rms_sample = np.sqrt(np.mean(v**2))
    rows.append({
        "T (K)": T,
        "sample mean": v.mean(),
        "theory v_mean": v_mean_theory(T),
        "SE": se,
        "z": (v.mean() - v_mean_theory(T)) / se,
        "sample v_rms": v_rms_sample,
        "theory v_rms": v_rms_theory(T),
        "v_rms error (%)": 100 * (v_rms_sample / v_rms_theory(T) - 1),
    })

pd.DataFrame(rows).set_index("T (K)").round(2)
```

- `mb.groupby("Temperature_K")`: loops over the four temperatures. `T` is the temperature and `g` holds its 10,000 rows.
- `se = v.std() / np.sqrt(len(v))`: the standard error. It is the typical size of the random error in a mean of 10,000 values.
- `z`: how many standard errors the sample mean is away from the theoretical mean.
- `np.sqrt(np.mean(v**2))`: the sample root-mean-square speed. It is the same definition as $v_{rms}$, applied to data.
- **Result:** all four $|z|$ values are below 2 (0.87, 0.04, -1.80 and 0.26). The gaps between the sample and theory are the size we expect from random sampling.
- 500 K has the largest gap (-4.6 m/s, $z = -1.80$). It is still inside the ±2 range, so it is not a problem. This seed simply gave slightly slow molecules at 500 K.
- The sample $v_{rms}$ is within 1% of theory at every temperature (largest error -0.80% at 500 K).
- The SE grows with temperature (1.16 to 3.71 m/s), because hotter gases have more spread. In relative terms it is always about 0.42%, the same CV for every temperature.
- **Conclusion:** the simulated data agrees with the Maxwell-Boltzmann theory. We can use `mb` for the analysis.

## Physical columns: speed squared and kinetic energy

The kinetic energy of one molecule is

$$E_k = \tfrac{1}{2} m v^2$$

The equipartition theorem says the **average** kinetic energy depends only on the temperature, not on the type of gas:

$$\langle E_k \rangle = \tfrac{3}{2} k_B T$$

The $3$ comes from the three directions of motion ($x$, $y$, $z$), and each direction carries $\tfrac{1}{2}k_BT$ on average.

```python
mb
```

```python
mb["Speed_Squared"] = mb["Speed"] ** 2                          # m^2/s^2
mb["Kinetic_Energy"] = 0.5 * M_MOLECULE * mb["Speed_Squared"]   # J

energy = mb.groupby("Temperature_K")["Kinetic_Energy"].mean().to_frame("mean E_k (J)")
energy["3/2 k_B T (J)"] = 1.5 * K_B * energy.index
energy["mean E_k / (k_B T)"] = energy["mean E_k (J)"] / (K_B * energy.index)

print(mb.head(3), "\n")
energy.style.format({"mean E_k (J)": "{:.4e}", "3/2 k_B T (J)": "{:.4e}", "mean E_k / (k_B T)": "{:.4f}"})
```

- `mb["Speed_Squared"]`: $v^2$ for every molecule, in m²/s².
- `mb["Kinetic_Energy"]`: $\tfrac{1}{2} m v^2$ in joules. The values are tiny (about $10^{-21}$ J) because one molecule is very light.
- `energy.index`: the four temperatures, so `1.5 * K_B * energy.index` gives $\tfrac{3}{2}k_BT$ for each row.
- `mean E_k / (k_B T)`: the energy measured in units of $k_BT$. Theory says it should be **1.5** at every temperature.
- **Result:** the ratios are 1.5108, 1.5076, 1.4760 and 1.5076. All of them are close to 1.5, so equipartition holds in our data.
- The energy errors (+0.72%, +0.51%, -1.60%, +0.51%) are exactly **twice** the $v_{rms}$ errors from the previous cell, because $\langle E_k \rangle = \tfrac{1}{2} m\, v_{rms}^2$. A squared quantity doubles the relative error.
- 500 K is again the furthest from theory, for the same reason as before (this sample has slightly slow molecules).
- **Note for later:** `Speed_Squared` and `Kinetic_Energy` are calculated only from `Speed`. They add **no new information**, so we will not use all three as separate model features.

## Theory columns, normalised speed and data leakage

We add the theoretical speeds for each molecule's temperature, plus the **normalised speed**:

$$u = \frac{v}{v_{mp}(T)}$$

These columns are useful for checking the physics. But they are calculated **from the temperature itself**, so a model that sees them would already know the answer. This is called **data leakage**. They must **never** be model inputs.

```python
mb["V_mp_theory"] = v_most_probable(mb["Temperature_K"])
mb["V_mean_theory"] = v_mean_theory(mb["Temperature_K"])
mb["V_rms_theory"] = v_rms_theory(mb["Temperature_K"])
mb["Normalised_Speed"] = mb["Speed"] / mb["V_mp_theory"]

LEAKAGE_COLS = ["V_mp_theory", "V_mean_theory", "V_rms_theory", "Normalised_Speed"]   # never use as model inputs

summary = mb.groupby("Temperature_K").agg(
    V_mp_unique=("V_mp_theory", "nunique"),
    V_mp_theory=("V_mp_theory", "first"),
    Norm_mean=("Normalised_Speed", "mean"),
    Norm_std=("Normalised_Speed", "std"),
)
print("columns:", list(mb.columns))
print(f"theory for u: mean = {2 / np.sqrt(np.pi):.4f}, std = {np.sqrt(1.5 - 4 / np.pi):.4f}\n")
summary.round(4)
```

- `v_most_probable(mb["Temperature_K"])`: our functions also work on a whole column, so every row gets the theory value for its own temperature.
- `Normalised_Speed`: each speed divided by $v_{mp}$ of its temperature. $u = 1$ means "exactly at the peak".
- `LEAKAGE_COLS`: a list we will use later to **remove** these columns before any model is trained.
- `.agg(...)`: several summaries at once. `nunique` counts different values, and `first` takes the first value.
- **Result 1 (leakage):** `V_mp_unique = 1` for every temperature. Each theory column has exactly **one value per class**, so it works as a perfect lookup table for the temperature. A model would score 100% accuracy and learn nothing about the physics.
- **Result 2 (universal curve):** the mean of $u$ is about 1.13 and its standard deviation about 0.48 at **all four** temperatures. Theory gives 1.1284 and 0.4762. After normalising, every temperature has the **same distribution**.
- `Normalised_Speed` is a hidden leak: it looks like a measurement, but it needs $T$ to be calculated. Dividing by $v_{mp}(T)$ removes exactly the temperature information we want the model to find.
- `Temperature_K` is the **target** (the label). `Speed` is the real measured feature.

## Outliers with the IQR rule

The IQR rule marks a value as an outlier if it lies outside the **fences**:

$$IQR = Q_3 - Q_1, \qquad \text{lower fence} = Q_1 - 1.5\,IQR, \qquad \text{upper fence} = Q_3 + 1.5\,IQR$$

We apply it **separately for each temperature**, because a speed that is extreme at 100 K is normal at 1000 K. Then we compare the outlier percentage with what the Maxwell-Boltzmann theory predicts for its own tail.

```python
rows = []
for T, g in mb.groupby("Temperature_K"):
    v = g["Speed"]
    q1, q3 = v.quantile([0.25, 0.75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    dist = stats.maxwell(loc=0, scale=maxwell_scale(T))
    rows.append({
        "T (K)": T,
        "lower fence": low,
        "upper fence": high,
        "below lower": (v < low).sum(),
        "above upper": (v > high).sum(),
        "outliers (%)": 100 * ((v < low) | (v > high)).mean(),
        "theory above upper (%)": 100 * dist.sf(high),
    })

pd.DataFrame(rows).set_index("T (K)").round(2)
```

- `v.quantile([0.25, 0.75])`: the first quartile $Q_1$ and the third quartile $Q_3$ of the speeds at one temperature.
- `low, high`: the IQR fences, calculated separately for each temperature.
- `(v < low) | (v > high)`: `True` for an outlier. `|` means "or". `.mean()` of True/False values gives the fraction of outliers.
- `dist.sf(high)`: the **survival function**, $P(v > \text{upper fence}) = 1 - \text{cdf}$. It is the fraction of molecules the theory **expects** above the fence.
- **Result 1:** the lower fence is **negative** at every temperature, so no molecule can be below it. All outliers are on the fast side, because of the long right tail.
- **Result 2:** about 0.8% to 1.1% of the molecules are above the upper fence (76 to 107 per temperature). The theory predicts 0.8% to 1.0%, so the outliers are **expected by the physics**.
- **Decision: keep all outliers.** They are not errors. They are real fast molecules, and they are exactly the tail that makes Maxwell-Boltzmann asymmetric. Removing them would make the mean, $v_{rms}$ and the kinetic energy too low, and the data would no longer follow the theory.
- The IQR rule assumes a roughly **symmetric** distribution. For a skewed distribution it always flags part of the normal tail, so it is a check here, not a cleaning rule.

## EDA: descriptive statistics by temperature

This starts the **Data Understanding** (EDA) phase. For each temperature we calculate the centre (mean, median), the spread (standard deviation) and the **shape**:

- **Skewness:** asymmetry. 0 = symmetric, > 0 = long tail on the right.
- **Excess kurtosis:** how heavy the tails are compared with a normal distribution (normal = 0).
- **Coefficient of variation:** $CV = s / \bar{v}$, the spread relative to the mean.

Because temperature only **stretches** the curve (the `scale`), theory says these three shape numbers are the **same at every temperature**: skewness 0.4857, excess kurtosis 0.1082, CV 0.4220.

```python
desc = mb.groupby("Temperature_K")["Speed"].agg(
    mean="mean",
    median="median",
    std="std",
    skewness="skew",
    excess_kurtosis=lambda v: v.kurt(),
)
desc["CV"] = desc["std"] / desc["mean"]

d0 = stats.maxwell()   # standard shape (scale = 1); the shape numbers do not depend on scale
print(f"theory for every T: skewness = {float(d0.stats(moments='s')):.4f}, "
      f"excess kurtosis = {float(d0.stats(moments='k')):.4f}, CV = {d0.std() / d0.mean():.4f}\n")
desc.round(4)
```

- `.agg(mean="mean", ...)`: named aggregation again. The left name is the new column, the right side is the calculation.
- `lambda v: v.kurt()`: a small one-line function. pandas' `kurt()` gives the **excess** kurtosis (normal = 0).
- `stats.maxwell()`: the standard Maxwell shape with scale = 1. `.stats(moments="s")` gives its skewness and `moments="k"` its excess kurtosis.
- **Centre and spread grow with T:** the mean goes from 275.9 to 870.3 m/s, and the std from 116.4 to 370.7 m/s. Both grow by about $\sqrt{10} \approx 3.16$ from 100 K to 1000 K.
- **Mean > median at every temperature** (for example 476.3 vs 460.3 m/s at 300 K). The long right tail pulls the mean up. This is what a positive skew looks like.
- **Shape stays the same:** skewness 0.46 to 0.52 (theory 0.4857), CV 0.420 to 0.429 (theory 0.4220), and excess kurtosis 0.04 to 0.24 (theory 0.1082). The values go up and down **with no trend** in T. The differences are random sampling noise.
- Kurtosis is the noisiest statistic. It depends on $v^4$, so a few very fast molecules in the tail change it a lot.
- **Important for RQ1:** skewness, kurtosis and CV do **not** change with temperature, so they **cannot help** a model to classify temperature. Only the **scale** statistics (mean, median, std) carry temperature information.

## EDA: mean speed and v_rms against T and √T

The theory says every characteristic speed is proportional to $\sqrt{T}$:

$$v_{mean} = \sqrt{\frac{8k_B}{\pi m}}\,\sqrt{T}, \qquad v_{rms} = \sqrt{\frac{3k_B}{m}}\,\sqrt{T}$$

So against $T$ the points should lie on a **curve**, but against $\sqrt{T}$ they should lie on a **straight line through the origin**. The slopes are fixed by physics. This is the idea we will use later for the Linear Regression.

```python
by_T = mb.groupby("Temperature_K")["Speed"].agg(mean="mean", rms=lambda v: np.sqrt(np.mean(v**2)))
by_T["sqrt_T"] = np.sqrt(by_T.index)
by_T["mean / sqrt(T)"] = by_T["mean"] / by_T["sqrt_T"]
by_T["rms / sqrt(T)"] = by_T["rms"] / by_T["sqrt_T"]

T_line = np.linspace(0, 1100, 300)
fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)
panels = [(axes[0], by_T.index, T_line, "temperature T (K)", "against T: a curve"),
          (axes[1], by_T["sqrt_T"], np.sqrt(T_line), "$\\sqrt{T}$ (K$^{1/2}$)", "against $\\sqrt{T}$: a straight line")]

for ax, x, x_line, xlabel, title in panels:
    ax.plot(x_line, v_mean_theory(T_line), color="tab:blue", label="theory $v_{mean}$")
    ax.plot(x_line, v_rms_theory(T_line), color="tab:red", ls="--", label="theory $v_{rms}$")
    ax.plot(x, by_T["mean"], "o", color="tab:blue", markersize=8, label="sample mean")
    ax.plot(x, by_T["rms"], "s", color="tab:red", markersize=8, label="sample $v_{rms}$")
    ax.set_xlabel(xlabel)
    ax.set_title(title)
    ax.legend()
axes[0].set_ylabel("speed (m/s)")

plt.tight_layout()
plt.savefig("figures/speed_vs_T_and_sqrtT.png", dpi=200)
plt.show()

print(f"theory slopes: v_mean / sqrt(T) = {v_mean_theory(1):.2f}, v_rms / sqrt(T) = {v_rms_theory(1):.2f}  m/s per K^0.5")
by_T.round(2)
```

- `by_T`: one row per temperature with the sample mean, the sample $v_{rms}$, $\sqrt{T}$, and each speed divided by $\sqrt{T}$.
- `panels = [...]`: a list with the settings of the two graphs. The `for` loop draws the same things twice: once against $T$ and once against $\sqrt{T}$.
- `v_mean_theory(T_line)`: the theory lines from Cell 4. In the right graph, the same values are plotted at `np.sqrt(T_line)`.
- `v_mean_theory(1)`: the theory value at $T = 1$ K is exactly the **slope** $\sqrt{8k_B/\pi m}$, because $\sqrt{1} = 1$.
- **Left graph:** against $T$, the points lie on a **curve** that rises quickly at first and then more slowly. Doubling $T$ does **not** double the speed.
- **Right graph:** against $\sqrt{T}$, the same points lie on **straight lines through the origin (0, 0)**. This confirms $v \propto \sqrt{T}$.
- **Slopes:** `mean / sqrt(T)` is 27.29 to 27.59 (theory 27.49), and `rms / sqrt(T)` is 29.60 to 29.92 (theory 29.84), in m/s per K$^{1/2}$. The values are almost constant, which is what "straight line through the origin" means.
- The $v_{rms}$ line is always above the mean line, by the fixed factor $\sqrt{3\pi/8} \approx 1.085$ (Cell 4).
- **For the Linear Regression later:** a straight line $\hat{y} = a \cdot x + b$ fits well if we use $x = \sqrt{T}$, but badly if we use $x = T$. Choosing the right feature from physics is **feature engineering**.
- The figure is saved as `figures/speed_vs_T_and_sqrtT.png`.

## EDA: the kinetic-energy distribution is a Gamma distribution

If we change the variable from speed to energy with $E = \tfrac{1}{2}mv^2$, the Maxwell-Boltzmann distribution becomes

$$f(E) = \frac{2}{\sqrt{\pi}}\,\frac{\sqrt{E}}{(k_BT)^{3/2}}\,e^{-E/k_BT}$$

This is a **Gamma distribution** with shape $\tfrac{3}{2}$ and scale $k_BT$. A Gamma(shape $k$, scale $\theta$) has mean $k\theta$ and variance $k\theta^2$, so

$$\langle E \rangle = \tfrac{3}{2}k_BT, \qquad \text{Var}(E) = \tfrac{3}{2}(k_BT)^2$$

If we measure the energy in units of $k_BT$, every temperature should follow the **same** curve, Gamma(3/2, 1), with mean 1.5 and variance 1.5.

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
e_line = np.linspace(0, 80, 500)    # energy in units of 1e-21 J
u_line = np.linspace(0, 8, 400)     # energy in units of k_B T
rows = []

for T in TEMPERATURES:
    E = mb.loc[mb["Temperature_K"] == T, "Kinetic_Energy"]
    kT = K_B * T

    axes[0].hist(E * 1e21, bins=80, range=(0, 80), density=True, color=TEMP_COLORS[T], alpha=0.35)
    axes[0].plot(e_line, stats.gamma.pdf(e_line, a=1.5, scale=kT * 1e21), color=TEMP_COLORS[T], lw=2, label=f"{T} K")

    axes[1].hist(E / kT, bins=80, range=(0, 8), density=True, histtype="step", color=TEMP_COLORS[T], lw=1.5, label=f"{T} K")
    rows.append({"T (K)": T, "mean E/kT": (E / kT).mean(), "variance E/kT": (E / kT).var()})

axes[1].plot(u_line, stats.gamma.pdf(u_line, a=1.5), color="black", ls="--", lw=2.5, label="Gamma(3/2, 1)")

axes[0].set_xlabel("kinetic energy E (10$^{-21}$ J)")
axes[0].set_ylabel("probability density")
axes[0].set_title("Energy in joules: one Gamma curve per T")
axes[1].set_xlabel("E / k$_B$T")
axes[1].set_title("Energy in units of k$_B$T: one curve for all T")
for ax in axes:
    ax.legend()

plt.tight_layout()
plt.savefig("figures/kinetic_energy_gamma.png", dpi=200)
plt.show()

print("theory: mean = 1.5, variance = 1.5")
pd.DataFrame(rows).set_index("T (K)").round(4)
```

- `E * 1e21`: the energies are about $10^{-21}$ J, so we multiply by $10^{21}$ to get readable numbers on the x-axis.
- `stats.gamma.pdf(x, a=1.5, scale=kT)`: the Gamma PDF. `a` is the **shape** (3/2) and `scale` is $k_BT$. It uses the same `loc` and `scale` idea as `maxwell`.
- `E / kT`: the energy in units of $k_BT$. This removes the temperature, like `Normalised_Speed` did for the speed (so it is for EDA only, never a model input).
- `histtype="step"`: draws only the outline of the histogram, so four histograms can be seen on top of each other.
- **Left graph:** in joules, each temperature has its own Gamma curve. Hotter gas has a lower, much wider curve with a long tail to high energies. At 100 K the bars are only 1 unit wide while the peak is very narrow, so they look a little lower than the curve.
- **Right graph:** in units of $k_BT$, all four histograms fall on **one** curve, Gamma(3/2, 1). This is the energy version of the universal curve from Cell 10.
- **Table:** the mean of $E/k_BT$ is 1.476 to 1.511 and the variance 1.467 to 1.555. Theory gives **1.5** for both. Equipartition (Cell 9) is confirmed, and so is the spread.
- **A surprise:** the most probable energy is at $E = \tfrac{1}{2}k_BT$ (the peak of the right graph), **not** at $\tfrac{1}{2} m v_{mp}^2 = k_BT$. The peak of a distribution moves when we change the variable from $v$ to $E = \tfrac{1}{2}mv^2$.
- **Why it is useful:** energy follows a well-known distribution, so we have a second, independent way to test the data. Later we can use this in a goodness-of-fit test.
- The figure is saved as `figures/kinetic_energy_gamma.png`.

## EDA: Q-Q plots against Maxwell-Boltzmann and against a normal distribution

A **Q-Q (quantile-quantile) plot** sorts the data and plots each sample quantile against the quantile the theory expects at the same position. For example, the 1st percentile of the data is plotted against the 1st percentile of the theory.

- If the data follows the theory, the points lie on the straight line $y = x$.
- If the points **bend away** from the line, the theory is wrong for this data, especially in the tails.

We compare two theories: the **Maxwell-Boltzmann** distribution, which should fit, and a **normal** distribution, which should not, because the data is skewed.

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
rows = []

for T in TEMPERATURES:
    speeds = mb.loc[mb["Temperature_K"] == T, "Speed"]

    (q_theory, q_sample), (slope, intercept, r) = stats.probplot(
        speeds, sparams=(0, maxwell_scale(T)), dist=stats.maxwell)
    axes[0].plot(q_theory, q_sample, ".", ms=2, color=TEMP_COLORS[T], label=f"{T} K")

    (q_norm, q_sample_n), (_, _, r_norm) = stats.probplot(speeds, dist="norm")
    z_sample = (q_sample_n - speeds.mean()) / speeds.std()   # standardised, so all T share one axis
    axes[1].plot(q_norm, z_sample, ".", ms=2, color=TEMP_COLORS[T], label=f"{T} K")

    rows.append({"T (K)": T, "r vs Maxwell": r, "slope vs Maxwell": slope, "r vs normal": r_norm})

lim = [0, 2700]
axes[0].plot(lim, lim, "k--", lw=1, label="y = x")
axes[0].set_xlabel("theoretical Maxwell-Boltzmann quantile (m/s)")
axes[0].set_ylabel("sample quantile (m/s)")
axes[0].set_title("Q-Q plot against Maxwell-Boltzmann")

axes[1].plot([-4, 4], [-4, 4], "k--", lw=1, label="y = x")
axes[1].set_xlabel("theoretical normal quantile (z)")
axes[1].set_ylabel("standardised sample quantile (z)")
axes[1].set_title("Q-Q plot against a normal distribution")

for ax in axes:
    ax.legend(markerscale=5)

plt.tight_layout()
plt.savefig("figures/qq_plots.png", dpi=200)
plt.show()

pd.DataFrame(rows).set_index("T (K)").round(4)
```

- `stats.probplot(speeds, sparams=(0, maxwell_scale(T)), dist=stats.maxwell)`: calculates the Q-Q points. `sparams` are the `loc` and `scale` of the theory, so the theory quantiles are in m/s.
- It returns two things: the points `(q_theory, q_sample)`, and a fitted straight line `(slope, intercept, r)`. `r` is the correlation of the points with that line (1 = perfectly straight).
- `dist="norm"`: the same plot against a normal distribution. Each sample is standardised to $z = (v - \bar{v})/s$, so all four temperatures share one axis.
- `_`: a Python name for "a value we do not need".
- **Left graph (Maxwell-Boltzmann):** the points of all four temperatures lie on $y = x$ from 0 to about 2,000 m/s. The slopes are 0.989 to 1.016 and $r \geq 0.9998$. The data follows the theory.
- At the very top a few points scatter around the line. These are the 10 to 20 fastest molecules out of 10,000. Extreme quantiles always have large random errors, so this is expected.
- **Right graph (normal):** the points make a clear **curve** (a "banana" shape), not a straight line.
- At the **bottom** the points flatten at about $z = -2.2$. A speed cannot go below 0, and $0$ m/s is only $1/CV \approx 2.37$ standard deviations below the mean. A normal distribution would allow negative speeds.
- At the **top** the points rise above the line: the data has a **longer right tail** than a normal distribution (positive skew).
- **Note:** $r$ against the normal is still 0.992. A high $r$ alone does not prove a good fit, because the bend is clear in the plot. This is why we use a formal test (Shapiro-Wilk) next.
- The figure is saved as `figures/qq_plots.png`.

A heatmap shows $r$ for every pair of columns at once. We use it to see three things: how strongly `Speed` is related to the temperature, which columns are **redundant**, and which columns **leak** the temperature.

- `mb[corr_cols].corr(method="pearson")`: the correlation of every pair of columns, as an 8 × 8 table.
- `sns.heatmap(..., annot=True, fmt=".2f")`: colours each cell by its value and writes the number with 2 decimals.
- `cmap="RdBu_r", vmin=-1, vmax=1`: red for positive, blue for negative, white for 0. Fixing the limits at -1 and 1 makes the colours easy to compare.
- **Speed vs temperature: r = 0.64.** This is a clear but only **moderate** relationship. $r^2 = 0.41$, so the temperature explains only about 41% of the variation in a single molecule's speed. The rest is the random spread **inside** each temperature. This is why one molecule is a weak clue for the temperature.
- **Redundancy:** `Speed_Squared` and `Kinetic_Energy` have r = 1.00 with each other (one is the other times $\tfrac{1}{2}m$) and r = 0.95 with `Speed`. Pearson is 0.95 and not 1 only because $v^2$ is a **curve** of $v$. Spearman gave 1.0 in Cell 14.
- **Leakage (theory columns):** r = 0.99 with the temperature, and 1.00 with each other. They are not exactly 1 with $T$ only because they follow $\sqrt{T}$, which is a curve. Each one is a perfect lookup table for the class (Cell 10).

Together with `Speed`, it gives the temperature exactly: $T = \frac{m}{2k_B}\left(\frac{v}{u}\right)^2$. Correlation cannot find this kind of leak. Only knowing **how a column was made** can.
- **Conclusion of the EDA:** `Speed` is the only honest feature for a single molecule, and it carries only moderate information about $T$.
- The figure is saved as `figures/correlation_heatmap.png`.

## Statistical test 1: Shapiro-Wilk test for normality

The Shapiro-Wilk test checks whether data comes from a **normal** distribution:

- $H_0$: the data is normal.
- $H_1$: the data is not normal.
- If $p < 0.05$ we **reject** $H_0$. The statistic $W$ is close to 1 for normal data.

We test two things at each temperature:

1. **500 single speeds.** The Maxwell-Boltzmann distribution is skewed, so we expect **not normal**. (We use 500 and not 10,000, because with a very large $n$ the test rejects even tiny, unimportant differences. SciPy also warns that its $p$-value is less accurate above $n = 5000$.)
2. **200 sample means**, each the mean of 50 molecules. The **Central Limit Theorem** says that means of many values are close to normal, even when the single values are not. So we expect **normal**.

```python
ALPHA = 0.05
rows = []

for T in TEMPERATURES:
    speeds = mb.loc[mb["Temperature_K"] == T, "Speed"]

    single = speeds.sample(n=500, random_state=SEED)                  # 500 single molecules
    means = speeds.to_numpy().reshape(200, 50).mean(axis=1)           # 200 means of 50 molecules

    w1, p1 = stats.shapiro(single)
    w2, p2 = stats.shapiro(means)
    rows.append({
        "T (K)": T,
        "W (single speeds)": w1, "p (single speeds)": p1,
        "normal? (single)": "yes" if p1 >= ALPHA else "no",
        "W (sample means)": w2, "p (sample means)": p2,
        "normal? (means)": "yes" if p2 >= ALPHA else "no",
    })

shapiro_table = pd.DataFrame(rows).set_index("T (K)")
shapiro_table.style.format({"W (single speeds)": "{:.4f}", "p (single speeds)": "{:.2e}",
                            "W (sample means)": "{:.4f}", "p (sample means)": "{:.3f}"})
```

- `ALPHA = 0.05`: the significance level. We reject $H_0$ when $p < 0.05$.
- `speeds.sample(n=500, random_state=SEED)`: picks 500 random molecules. The seed makes it the same 500 every time.
- `speeds.to_numpy().reshape(200, 50)`: puts the 10,000 speeds into a table of 200 rows × 50 columns. `.mean(axis=1)` takes the mean of each row, giving 200 sample means. Every molecule is used exactly once.
- `stats.shapiro(x)`: returns the statistic $W$ and the $p$-value.
- **Single speeds:** $p$ is between $4 \times 10^{-6}$ and $2 \times 10^{-4}$ at all four temperatures, far below 0.05. We **reject normality**. This agrees with the skewness (Cell 15) and the bent normal Q-Q plot (Cell 20).
- $W$ is still about 0.98, which looks close to 1. As with $r$ in the Q-Q plot, a number close to 1 does not mean "normal". The $p$-value decides.
- **Sample means:** $p$ is between 0.057 and 0.424, all above 0.05. We **cannot reject normality**. The means of 50 skewed values already behave like a normal distribution: this is the **Central Limit Theorem**.
- At 1000 K, $p = 0.057$ is only just above 0.05. Means of 50 molecules still keep a small skewness, about $0.49/\sqrt{50} \approx 0.07$, so they are only **approximately** normal. Larger groups would make them more normal.
- **Why this matters:** methods that assume normality (for example a t-test or a confidence interval for the mean) are fine for **sample means**, but not for **single speeds**. For single speeds, the Q-Q plot (Cell 20) and the histograms (Cell 16) already show that Maxwell-Boltzmann is the correct model.

## Machine Learning 1: Linear Regression. Building the regression dataset

**Goal:** predict the **mean speed** of a gas from its temperature, and check whether the model learns the physics $v_{mean} = 27.49\,\sqrt{T}$.

Four temperatures are too few points for a regression, so we use the **91 temperatures of the NIST table** (`df`, 100 to 1000 K in steps of 10 K). For each temperature we simulate 1,000 molecules and take their **sample mean speed**. This is our target $y$. It contains realistic random noise.

To test **extrapolation**, we split by temperature instead of randomly:

- **Train:** $T \leq 700$ K (61 temperatures).
- **Test:** $T > 700$ K (30 temperatures). The model never sees these hot temperatures during training.

```python
N_REG = 1_000                                  # molecules per temperature
rng_reg = np.random.default_rng(SEED + 1)      # separate generator: independent of mb
reg_T = np.sort(df["Temperature (K)"].unique())

reg = pd.DataFrame({"T": reg_T, "sqrt_T": np.sqrt(reg_T)})
reg["mean_speed"] = [stats.maxwell.rvs(loc=0, scale=maxwell_scale(T), size=N_REG, random_state=rng_reg).mean()
                     for T in reg_T]
reg["theory_mean"] = v_mean_theory(reg["T"])   # for comparison only, never a feature
reg["split"] = np.where(reg["T"] <= 700, "train", "test")

print("shape:", reg.shape)
print(reg["split"].value_counts().to_string(), "\n")
reg.iloc[[0, 1, 2, 60, 61, 90]].round(2)
```

## Machine Learning 1: Linear Regression. Fitting T vs $\sqrt{T}$ and testing extrapolation

We fit two **Linear Regression** models on the train set ($T \leq 700$ K) and test them on the hot gas ($T > 700$ K):

- **Model 1 (naive):** $\hat{v} = b_0 + b_1\,T$. It assumes speed grows linearly with temperature.
- **Model 2 (physics):** $\hat{v} = b_0 + b_1\,\sqrt{T}$. Theory says $b_0 = 0$ and $b_1 = \sqrt{8k_B/\pi m} = 27.49$ m/s per K$^{1/2}$.

Evaluation metrics on the test set:

$$R^2 = 1 - \frac{\sum (y - \hat{y})^2}{\sum (y - \bar{y})^2} \qquad RMSE = \sqrt{\frac{1}{n}\sum (y - \hat{y})^2}$$

$R^2$ close to 1 is good; RMSE is the typical error in m/s. Because the target itself has about 1.3% random noise, a perfect model cannot have RMSE = 0.

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

train, test = reg[reg["split"] == "train"], reg[reg["split"] == "test"]
models, results = {}, []
for feature in ["T", "sqrt_T"]:
    model = LinearRegression().fit(train[[feature]], train["mean_speed"])
    pred = model.predict(test[[feature]])
    models[feature] = model
    reg["pred_" + feature] = model.predict(reg[[feature]])
    results.append({"feature": feature, "intercept": model.intercept_, "slope": model.coef_[0],
                    "train_R2": model.score(train[[feature]], train["mean_speed"]),
                    "test_R2": r2_score(test["mean_speed"], pred),
                    "test_RMSE": np.sqrt(mean_squared_error(test["mean_speed"], pred))})
print("theory slope for sqrt_T:", round(float(np.sqrt(8 * constants.k / (np.pi * M_MOLECULE))), 2))
print("RMSE of theory itself on test:",
      round(float(np.sqrt(mean_squared_error(test["mean_speed"], test["theory_mean"]))), 2))
```

```python

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), sharey=True)
for ax, feature in zip(axes, ["T", "sqrt_T"]):
    ax.scatter(train[feature], train["mean_speed"], s=12, color="tab:blue", label="train (T <= 700 K)")
    ax.scatter(test[feature], test["mean_speed"], s=12, color="tab:red", label="test (T > 700 K)")
    ax.plot(reg[feature], reg["pred_" + feature], color="black", label="fitted line")
    ax.axvline(reg.loc[reg["T"] == 700, feature].iloc[0], color="grey", linestyle="--")
    ax.set_xlabel("T (K)" if feature == "T" else "sqrt(T) (K^0.5)")
    ax.set_title(f"Model with x = {feature}")
    ax.legend()
axes[0].set_ylabel("mean speed (m/s)")
plt.tight_layout()
plt.savefig("figures/regression_T_vs_sqrtT.png", dpi=200)
plt.show()

pd.DataFrame(results).round(4)
```

- `train[[feature]]`: double brackets give a **DataFrame** (a table with one column). scikit-learn always expects a 2-D input `X`, even with only one feature.
- `LinearRegression().fit(X, y)`: finds the intercept $b_0$ and slope $b_1$ that minimise the squared errors (Ordinary Least Squares). The model only sees the **train** rows.
- `model.score(...)` gives $R^2$ on the train set. `r2_score` and `mean_squared_error` measure the **test** set (710 to 1000 K), which the model never saw.
- `RMSE of theory itself`: the error of the exact formula $v_{mean}$ against the noisy sample means. This is the **best possible RMSE**: no model can beat the random noise in the target.
- **Model 1, x = T:** train $R^2$ = 0.9834 looks very good, but test $R^2$ = **−1.60** and RMSE = **66.4 m/s**. A negative $R^2$ means the model is worse than simply predicting the average of the test set. The straight line in $T$ keeps rising too fast, while the real speeds bend down (left plot). The intercept 244.5 m/s also has no physical meaning: at 0 K the speed must be 0.
- **Model 2, x = √T:** train $R^2$ = 0.9965, test $R^2$ = **0.931** and RMSE = **10.8 m/s**, which is about the noise level (theory itself: 11.06 m/s). So the model is as good as the physics formula.
- **Learned parameters:** slope = **27.39** vs theory **27.49** m/s per K$^{1/2}$ (0.4% difference), and intercept = **0.81 m/s** vs theory 0. The model has rediscovered $v_{mean} = \sqrt{8k_BT/\pi m}$ from data only.
- **Why is test $R^2$ 0.93 and not 0.99?** The test range (710 to 1000 K) is narrow, so the true variation is small compared with the random noise of about 1.3%. RMSE is the fairer metric here.
- **Lesson:** a high train $R^2$ does not prove a good model. Choosing the correct feature ($\sqrt{T}$, from physics) is what allows safe **extrapolation**. Figure saved as `figures/regression_T_vs_sqrtT.png`.

## Validation with real data: NIST speed of sound vs the regression model

So far all speeds were **simulated**. The NIST table (Part 1) contains a **real measured property**: the speed of sound $c$ in N$_2$. Kinetic theory links sound to the molecular speeds:

$$c = \sqrt{\frac{\gamma\,k_B T}{m}} \qquad v_{mean} = \sqrt{\frac{8\,k_B T}{\pi m}} \qquad \Rightarrow \qquad \frac{c}{v_{mean}} = \sqrt{\frac{\gamma\,\pi}{8}}$$

where $\gamma = C_p / C_v$ is the heat capacity ratio (also from NIST). For an ideal diatomic gas $\gamma = 7/5 = 1.4$, so $c / v_{mean} = 0.742$: sound travels a bit slower than the average molecule.

**Test:** take the mean speed predicted by the $\sqrt{T}$ regression model, multiply by $\sqrt{\gamma\pi/8}$, and compare with the NIST speed of sound at all 91 temperatures.

```python
nist = df[["Temperature (K)", "Cp (J/mol*K)", "Cv (J/mol*K)", "Sound Spd. (m/s)"]].copy()
nist.columns = ["T", "Cp", "Cv", "c_nist"]
nist["gamma"] = nist["Cp"] / nist["Cv"]
nist["ratio_nist"] = nist["c_nist"] / v_mean_theory(nist["T"])     # measured c / v_mean
nist["ratio_theory"] = np.sqrt(nist["gamma"] * np.pi / 8)          # kinetic theory
v_model = models["sqrt_T"].predict(pd.DataFrame({"sqrt_T": np.sqrt(nist["T"])}))
nist["c_model"] = v_model * nist["ratio_theory"]                   # sound speed from the ML model
nist["error_%"] = (nist["c_model"] - nist["c_nist"]) / nist["c_nist"] * 100

print("max |error| (%):", round(nist["error_%"].abs().max(), 2))
print("mean |error| (%):", round(nist["error_%"].abs().mean(), 2))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
axes[0].plot(nist["T"], nist["c_nist"], "o", ms=4, label="NIST (measured)")
axes[0].plot(nist["T"], nist["c_model"], "-", color="black", label="regression model x sqrt(gamma*pi/8)")
axes[0].set(xlabel="T (K)", ylabel="speed of sound (m/s)", title="Speed of sound in N2 at 1 atm")
axes[1].plot(nist["T"], nist["ratio_nist"], "o", ms=4, label="NIST c / v_mean")
axes[1].plot(nist["T"], nist["ratio_theory"], "-", color="black", label="sqrt(gamma*pi/8)")
axes[1].axhline(np.sqrt(1.4 * np.pi / 8), color="grey", linestyle="--", label="ideal diatomic (gamma = 1.4)")
axes[1].set(xlabel="T (K)", ylabel="c / v_mean", title="Ratio of sound speed to mean molecular speed")
for ax in axes:
    ax.legend()
plt.tight_layout()
plt.savefig("figures/sound_speed_check.png", dpi=200)
plt.show()

nist[nist["T"].isin([100, 300, 500, 700, 1000])].round(4)
```

- `df[[...]].copy()`: takes only the four NIST columns we need. `.copy()` makes a new table, so changing `nist` never changes `df`.
- `nist["gamma"] = Cp / Cv`: the **real** heat capacity ratio of N$_2$ at each temperature, measured by NIST.
- `ratio_nist`: the measured sound speed divided by the theoretical $v_{mean}$. `ratio_theory`: the kinetic-theory prediction $\sqrt{\gamma\pi/8}$.
- `models["sqrt_T"].predict(...)`: reuses the **trained** regression model from Cell 24. We only give it $\sqrt{T}$; it never saw any NIST sound data.
- `c_model = v_model * ratio_theory`: converts the predicted mean molecular speed into a predicted speed of sound.
- **Result 1, the model predicts real sound speed.** The mean absolute error is **0.32%** over all 91 temperatures. At 300 K: model 352.5 m/s vs NIST 353.2 m/s (−0.19%). At 1000 K: 629.1 vs 631.1 m/s (−0.32%). The small negative error comes from the learned slope 27.39, which is 0.4% below the theory value 27.49.
- **Result 2, $\gamma$ is not exactly 1.4.** NIST gives $\gamma$ = 1.4012 at 300 K (ideal diatomic), but 1.3411 at 1000 K. At high temperature the N$_2$ molecules start to **vibrate**, which adds energy storage and lowers $\gamma$. The ratio $c / v_{mean}$ falls from 0.742 to 0.726, and the formula $\sqrt{\gamma\pi/8}$ follows it very closely (right plot).
- **Result 3, the largest error is at 100 K (+1.96%).** Here N$_2$ is close to its boiling point (77 K), so it is **not an ideal gas** anymore. Molecules attract each other, and the ideal-gas formula $c = \sqrt{\gamma k_B T/m}$ becomes less accurate. This is a real limit of the Maxwell-Boltzmann ideal-gas model, not a coding error.
- **Conclusion:** a model trained only on **simulated** molecular speeds predicts a **real measured** property of nitrogen within 2% everywhere and within 0.33% from 200 K upwards. This links Part 1 (NIST data) and Part 2 (simulation). Figure saved as `figures/sound_speed_check.png`.

## Machine Learning 2: Random Forest classification. Level A (one molecule)

**RQ1:** can a model recognise the gas temperature from molecular speeds?

**Level A** is the hardest version: the model sees the speed of **one single molecule** (`FEATURES_A = ["Speed"]`) and must say whether it came from the gas at 100, 300, 500 or 1000 K.

The four Maxwell-Boltzmann curves **overlap** strongly (Cell 16). A speed of 500 m/s is possible at every temperature. So even a perfect model cannot be always right. The best possible rule is the **Bayes classifier**: choose the temperature whose curve is highest at that speed,

$$\hat{T}(v) = \arg\max_T \; f(v \mid T)$$

We compare three numbers:

- **Baseline:** always guess the same class, accuracy $= 1/4 = 25\%$.
- **Random Forest:** an ensemble of decision trees. Each tree learns speed thresholds, and the trees vote.
- **Bayes ceiling:** the theoretical maximum, computed from the exact PDFs (physics, not learning).

Split: 80% train, 20% test, **stratified** so each temperature keeps 25% of the rows.

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay

X_A, y = mb[FEATURES_A], mb["Class"]
X_train_A, X_test_A, y_train, y_test = train_test_split(X_A, y, test_size=0.2, stratify=y, random_state=SEED)

baseline = DummyClassifier(strategy="most_frequent").fit(X_train_A, y_train)
rf_A = RandomForestClassifier(n_estimators=200, min_samples_leaf=50, random_state=SEED, n_jobs=-1)
rf_A.fit(X_train_A, y_train)
pred_A = rf_A.predict(X_test_A)

pdfs = np.column_stack([maxwell_pdf(X_test_A["Speed"], T) for T in TEMPERATURES])  # one column per class
bayes_pred = pdfs.argmax(axis=1)                                                 # class with the highest curve

print("baseline accuracy :", round(accuracy_score(y_test, baseline.predict(X_test_A)), 4))
print("RF train accuracy :", round(rf_A.score(X_train_A, y_train), 4))
print("RF test accuracy  :", round(accuracy_score(y_test, pred_A), 4))
print("Bayes ceiling     :", round(accuracy_score(y_test, bayes_pred), 4))

ConfusionMatrixDisplay.from_predictions(y_test, pred_A, normalize="true", values_format=".2f",
                                        display_labels=[f"{T} K" for T in TEMPERATURES], cmap="Blues")
plt.title("Random Forest Level A (one molecule): row = true T")
plt.savefig("figures/rf_levelA_confusion.png", dpi=200, bbox_inches="tight")
plt.show()
```

## Machine Learning 2: Random Forest classification. Level B (samples of n molecules)

In a real experiment we never measure only one molecule. We measure **many** molecules and look at their statistics. In Level B the model receives a **sample of $n$ molecules** from the same gas, described by 8 statistics:

- `mean`, `std`, `median`, and the percentiles `q10`, `q25`, `q75`, `q90` of the speeds.
- `mean_sq` $= \frac{1}{n}\sum v^2$. Physics says $\langle v^2 \rangle = 3k_BT/m$, so this statistic is directly proportional to $T$.

**Physics expectation.** The random error of a sample mean shrinks like $1/\sqrt{n}$ (Central Limit Theorem). So the four temperature "clouds" overlap less and less as $n$ grows, and accuracy should rise towards 100%.

**Bayes ceiling for $n$ molecules.** For the Maxwell-Boltzmann distribution, the best possible rule only needs $\overline{v^2}$ (it is a **sufficient statistic**):

$$\hat{T} = \arg\max_T \left[ -\tfrac{3}{2}\ln a_T^2 \;-\; \frac{\overline{v^2}}{2\,a_T^2} \right], \qquad a_T^2 = \frac{k_BT}{m}$$

**No leakage:** train samples are built **only** from the train molecules of Cell 26, and test samples **only** from the test molecules. No molecule is in both.

```python
FEATURES_B = ["mean", "std", "q10", "q25", "median", "q75", "q90", "mean_sq"]

def make_samples(speeds, classes, n, seed=SEED):
    """Cut the molecules of each class into groups of n and describe every group with statistics."""
    rng_s = np.random.default_rng(seed)
    parts = []
    for c in np.unique(classes):
        v = rng_s.permutation(speeds[classes == c])
        groups = v[: len(v) // n * n].reshape(-1, n)          # one row = one sample of n molecules
        q10, q25, q50, q75, q90 = np.percentile(groups, [10, 25, 50, 75, 90], axis=1)
        parts.append(pd.DataFrame({"mean": groups.mean(axis=1), "std": groups.std(axis=1),
                                   "q10": q10, "q25": q25, "median": q50, "q75": q75, "q90": q90,
                                   "mean_sq": (groups**2).mean(axis=1), "Class": c}))
    return pd.concat(parts, ignore_index=True)

def bayes_predict(mean_sq):
    """Exact physics classifier: log-likelihood per molecule, using the sufficient statistic mean(v^2)."""
    a2 = np.array([maxwell_scale(T) ** 2 for T in TEMPERATURES])
    loglik = -1.5 * np.log(a2) - np.asarray(mean_sq)[:, None] / (2 * a2)
    return loglik.argmax(axis=1)

N_LIST = [1, 2, 5, 10, 20, 50, 100]
samples, rows = {}, []
for n in N_LIST:
    tr = make_samples(X_train_A["Speed"].to_numpy(), y_train.to_numpy(), n)   # train molecules only
    te = make_samples(X_test_A["Speed"].to_numpy(), y_test.to_numpy(), n)     # test molecules only
    samples[n] = (tr, te)
    rf = RandomForestClassifier(n_estimators=200, min_samples_leaf=20, random_state=SEED, n_jobs=-1)
    rf.fit(tr[FEATURES_B], tr["Class"])
    rows.append({"n": n, "train_samples": len(tr), "test_samples": len(te),
                 "RF_test_acc": accuracy_score(te["Class"], rf.predict(te[FEATURES_B])),
                 "Bayes_ceiling": accuracy_score(te["Class"], bayes_predict(te["mean_sq"]))})
levelB = pd.DataFrame(rows)

plt.figure(figsize=(7, 4.5))
plt.plot(levelB["n"], levelB["RF_test_acc"], "o-", label="Random Forest (test)")
plt.plot(levelB["n"], levelB["Bayes_ceiling"], "s--", color="black", label="Bayes ceiling (physics)")
plt.axhline(0.25, color="grey", linestyle=":", label="baseline 25%")
plt.xscale("log")
plt.xticks(N_LIST, N_LIST)
plt.xlabel("molecules per sample, n (log scale)")
plt.ylabel("accuracy")
plt.title("Level B: accuracy grows with the number of molecules")
plt.legend()
plt.savefig("figures/levelB_accuracy_vs_n.png", dpi=200, bbox_inches="tight")
plt.show()

levelB.round(4)
```

- `make_samples(...)`: for each class it shuffles the molecules (`permutation`), cuts them into groups of $n$ with `reshape(-1, n)`, and computes the 8 statistics for each group (one row per group). Leftover molecules that do not fill a full group are dropped.
- `np.percentile(groups, [10, 25, 50, 75, 90], axis=1)`: five percentiles for every row at once, without a loop.
- `bayes_predict(mean_sq)`: the formula above, computed for the 4 temperatures at once (`[:, None]` turns the column into a table of 4 columns). `argmax` picks the most likely temperature.
- The loop trains a new Random Forest for each $n$. `samples[n]` keeps the train and test tables, so the next cells can reuse them.
- **Result:**

| $n$ molecules | 1 | 2 | 5 | 10 | 20 | 50 | 100 |
|---|---|---|---|---|---|---|---|
| Random Forest | 49.1% | 59.3% | 75.2% | 87.3% | 94.8% | 99.4% | 100% |
| Bayes ceiling | 50.5% | 61.0% | 75.1% | 87.3% | 94.8% | 99.4% | 100% |

- **Physics view:** accuracy grows quickly because averaging removes randomness. With 50 molecules, the error of the mean speed is about $42\% / \sqrt{50} \approx 6\%$, while 300 K and 500 K differ by about 29% in mean speed. The clouds no longer overlap, so the temperature is almost always clear.
- **Data analyst view:** from $n = 5$ upwards, the Random Forest is **equal to the Bayes ceiling**. A model cannot do better than the physics allows, and this model reaches that limit. This is strong evidence that the model is not overfitting and the pipeline has no leakage (a leak would give results **above** the ceiling).
- **Answer to RQ1:** yes, temperature can be classified from speed distributions, **if** the model sees a sample. One molecule gives about 50%, 20 molecules give about 95%, and 100 molecules give 100%. Figure saved as `figures/levelB_accuracy_vs_n.png`.

## Model comparison with cross-validation (n = 5 molecules)

Is the Random Forest the best choice? We compare four different types of classifier. We use $n = 5$, because there the task is still hard (Bayes ceiling about 75%), so differences between models can be seen.

| Model | Idea |
|---|---|
| Logistic Regression | draws straight boundaries between classes (linear model) |
| k-Nearest Neighbours | looks at the 25 most similar training samples and takes a vote |
| Gaussian Naive Bayes | assumes each feature is normal inside each class |
| Random Forest | many decision trees vote |

**5-fold stratified cross-validation (CV)** on the train set: the train set is cut into 5 parts. Each model is trained on 4 parts and tested on the 5th, five times. The mean and the standard deviation of the 5 scores tell us how good and how **stable** the model is. The test set is used only once, at the end.

Logistic Regression and kNN use `StandardScaler` inside a **pipeline**, because they are sensitive to the size of the numbers. The pipeline fits the scaler **only** on the training folds, which avoids leakage during CV.

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import StratifiedKFold, cross_val_score

N_CMP = 5                                            # hard enough that models can differ
train_B, test_B = samples[N_CMP]
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
candidates = {
    "Logistic Regression": make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000)),
    "kNN (k=25)": make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=25)),
    "Gaussian Naive Bayes": GaussianNB(),
    "Random Forest": RandomForestClassifier(n_estimators=200, min_samples_leaf=20, random_state=SEED, n_jobs=-1),
}
rows = []
for name, model in candidates.items():
    scores = cross_val_score(model, train_B[FEATURES_B], train_B["Class"], cv=cv)   # train set only
    model.fit(train_B[FEATURES_B], train_B["Class"])
    rows.append({"model": name, "CV_mean": scores.mean(), "CV_std": scores.std(),
                 "test_acc": model.score(test_B[FEATURES_B], test_B["Class"])})
print("Bayes ceiling at n = 5:", round(accuracy_score(test_B["Class"], bayes_predict(test_B["mean_sq"])), 4))
comparison = pd.DataFrame(rows).sort_values("CV_mean", ascending=False)
comparison.round(4)
```

- `make_pipeline(StandardScaler(), model)`: first scales each feature to mean 0 and std 1, then fits the model. Inside `cross_val_score` the scaler is refitted in every fold.
- `StratifiedKFold(n_splits=5, shuffle=True)`: 5 folds, each with 25% of every temperature.
- `cross_val_score(...)`: returns the 5 fold accuracies. We report their mean and std.
- **Result (Bayes ceiling = 75.1%):**

| Model | CV mean | CV std | Test accuracy |
|---|---|---|---|
| Logistic Regression | 74.8% | 0.8% | 75.0% |
| Random Forest | 74.7% | 0.7% | 75.2% |
| Gaussian Naive Bayes | 74.4% | 0.9% | 74.6% |
| kNN (k = 25) | 74.2% | 0.5% | 75.0% |

- **Data analyst view:** all four models are within about 0.6 points of each other, and the differences are **smaller than the CV standard deviation**. So no model is really better. CV and test results agree, which shows the models generalise well.
- **Physics view:** all models hit the same wall, the Bayes ceiling of 75%. The limit is the **overlap of the physical distributions**, not the algorithm. When the data is this clean, a simple model (Logistic Regression) is as good as a complex one.
- We keep the **Random Forest** for the next cell because it gives feature importances, and it has the best test accuracy (75.2%).

## Evaluation of the Random Forest (n = 5): precision, recall and feature importance

Accuracy is one number for all classes. A senior analyst also checks **each class**:

$$\text{Precision} = \frac{TP}{TP + FP} \qquad \text{Recall} = \frac{TP}{TP + FN} \qquad F_1 = \frac{2 \cdot \text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$$

- **Precision:** when the model says "300 K", how often is it right?
- **Recall:** of all the real 300 K samples, how many did the model find?

**Feature importance** shows which statistics the trees use most (how much each feature reduces impurity in the splits).

```python
from sklearn.metrics import classification_report

best = candidates["Random Forest"]                    # already fitted on train_B in the previous cell
pred_B = best.predict(test_B[FEATURES_B])
print(classification_report(test_B["Class"], pred_B, target_names=[f"{T} K" for T in TEMPERATURES], digits=3))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
ConfusionMatrixDisplay.from_predictions(test_B["Class"], pred_B, normalize="true", values_format=".2f",
                                        display_labels=[f"{T} K" for T in TEMPERATURES], cmap="Blues", ax=axes[0])
axes[0].set_title(f"Random Forest, n = {N_CMP} molecules: row = true T")
importance = pd.Series(best.feature_importances_, index=FEATURES_B).sort_values()
importance.plot.barh(ax=axes[1], color="tab:blue")
axes[1].set(xlabel="importance (mean decrease in impurity)", title="Which statistics does the forest use?")
plt.tight_layout()
plt.savefig("figures/levelB_evaluation.png", dpi=200)
plt.show()

importance.sort_values(ascending=False).round(3)
```

- `classification_report(...)`: precision, recall and F1 for each class. `support` = number of test samples per class (400).
- `ConfusionMatrixDisplay(..., ax=axes[0])`: the confusion matrix in the left panel; `normalize="true"` shows proportions per true class.
- `best.feature_importances_`: one number per feature; they add up to 1.
- **Per-class results:**

| Class | Precision | Recall | F1 |
|---|---|---|---|
| 100 K | 0.902 | 0.925 | 0.914 |
| 300 K | 0.651 | 0.685 | 0.667 |
| 500 K | 0.597 | 0.578 | 0.587 |
| 1000 K | 0.859 | 0.820 | 0.839 |

- **Same pattern as Level A, but much better:** the extreme temperatures are easy (F1 0.91 and 0.84) and the middle ones are hard (F1 0.67 and 0.59), because 300 K and 500 K have neighbours on both sides. Macro and weighted averages are the same (0.752) because the classes are balanced.
- **Feature importance:** `mean_sq` (0.270) and `mean` (0.223) are the top two, then the upper percentiles `q75` (0.162) and `q90` (0.156). `std` is the least used (0.025).
- **Physics view:** the forest found by itself that $\overline{v^2}$ is the most useful statistic. This is exactly the **sufficient statistic** from the Bayes formula, and it is linked to temperature by equipartition: $\frac{1}{2}m\langle v^2 \rangle = \frac{3}{2}k_BT$. `std` is weak because with only 5 molecules it is a very noisy estimate.
- **Data analyst caution:** the 8 features are strongly correlated, so the importance is shared between them. The ranking is a guide, not an exact measure. Figure saved as `figures/levelB_evaluation.png`.

## Machine Learning 3: Clustering (K-Means and Hierarchical), without labels

Until now the models were told the correct temperature (**supervised** learning). Now we ask a different question: if we **hide the labels**, does the data form natural groups that match the four temperatures? (**Unsupervised** learning.)

We use the samples of $n = 50$ molecules (train + test together, 800 samples), standardised.

- **K-Means:** places $k$ centres and assigns each sample to the closest one. It minimises the **inertia** $= \sum \lVert x_i - \mu_{c(i)} \rVert^2$.
- **Elbow method:** plot inertia vs $k$ and look for the "bend".
- **Silhouette score** for each sample: $s = \frac{b - a}{\max(a, b)}$, where $a$ = mean distance to its own cluster and $b$ = mean distance to the nearest other cluster. Values close to 1 mean well-separated clusters.
- **Adjusted Rand Index (ARI):** compares the clusters with the true temperatures **after** clustering. 1 = perfect match, 0 = random.
- **Hierarchical clustering (Ward):** joins the closest groups step by step, as a second method to check K-Means.

```python
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score, adjusted_rand_score

N_CLU = 50
clu = pd.concat(samples[N_CLU], ignore_index=True)                 # labels are NOT used for fitting
X_clu = StandardScaler().fit_transform(clu[FEATURES_B])

K_RANGE = range(2, 9)
inertia, silhouette = [], []
for k in K_RANGE:
    km = KMeans(n_clusters=k, n_init=10, random_state=SEED).fit(X_clu)
    inertia.append(km.inertia_)
    silhouette.append(silhouette_score(X_clu, km.labels_))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(list(K_RANGE), inertia, "o-")
axes[0].set(xlabel="number of clusters k", ylabel="inertia (within-cluster SS)", title="Elbow method")
axes[1].plot(list(K_RANGE), silhouette, "o-", color="tab:orange")
axes[1].set(xlabel="number of clusters k", ylabel="silhouette score", title="Silhouette method")
plt.tight_layout()
plt.savefig("figures/kmeans_elbow_silhouette.png", dpi=200)
plt.show()

km4 = KMeans(n_clusters=4, n_init=10, random_state=SEED).fit(X_clu)
hc4 = AgglomerativeClustering(n_clusters=4, linkage="ward").fit(X_clu)
print("silhouette per k:", {k: round(float(s), 3) for k, s in zip(K_RANGE, silhouette)})
print("ARI K-Means (k=4)     :", round(adjusted_rand_score(clu["Class"], km4.labels_), 4))
print("ARI Hierarchical (k=4):", round(adjusted_rand_score(clu["Class"], hc4.labels_), 4), "\n")
pd.crosstab(clu["Class"].map(dict(enumerate(TEMPERATURES))), km4.labels_,
            rownames=["true T (K)"], colnames=["K-Means cluster"])
```

- `pd.concat(samples[N_CLU])`: joins the train and test samples. This is allowed here because clustering never uses the labels to fit.
- `StandardScaler().fit_transform(...)`: K-Means uses distances, so all features must have the same scale. Otherwise `mean_sq` (values around $10^5$) would dominate the others.
- `KMeans(n_clusters=k, n_init=10)`: runs K-Means 10 times from different starts and keeps the best one.
- `silhouette_score(X_clu, km.labels_)`: the average silhouette over all samples.
- **Silhouette per $k$:** 0.551 (2), **0.651 (3)**, 0.625 (4), 0.547 (5), then it falls below 0.48.
- **Honest finding:** the best silhouette is at $k = 3$, not 4. With $k = 3$, K-Means **joins 300 K and 500 K** into one cluster (198 + 198 samples). These two gases are the closest in speed, so without labels they look like one group. $k = 4$ is a very close second (0.625).
- **With $k = 4$:** ARI = **0.958** for K-Means and **0.967** for Hierarchical clustering. The table shows that 100 K and 1000 K are found almost perfectly (200 and 199 of 200). Only 12 samples of 300 K and 500 K are mixed up.
- **Physics view:** the data contains a natural structure that matches the temperatures, even with no labels. The groups are ordered by the speed scale $\sqrt{k_BT/m}$, and 300 K / 500 K are the closest pair because $\sqrt{500/300} = 1.29$ is the smallest ratio between neighbours.
- **Data analyst view:** the elbow and silhouette methods suggest 3 or 4 clusters, and domain knowledge (we have 4 temperatures) chooses 4. Two different algorithms agree (ARI 0.96 and 0.97), so the result is robust. Figure saved as `figures/kmeans_elbow_silhouette.png`.

## Machine Learning 4: PCA. Does dimensionality reduction keep the physics? (RQ2)

**Principal Component Analysis (PCA)** finds new axes (principal components) that are combinations of the original features. PC1 is the direction with the **largest variance**, PC2 the next one (at 90° to PC1), and so on:

$$PC_1 = w_1 x_1 + w_2 x_2 + \dots + w_8 x_8, \qquad \text{explained variance ratio} = \frac{\lambda_j}{\sum_k \lambda_k}$$

The weights $w$ are called **loadings**.

**Physics prediction.** Temperature only **stretches** the Maxwell-Boltzmann curve (one scale parameter $a = \sqrt{k_BT/m}$), while its **shape** does not change (Cell 15). So all 8 statistics should move together, and **one component** should hold almost all the information. That component should be proportional to $\sqrt{T}$.

**RQ2 test:**
1. How much variance does PC1 explain?
2. Are the PC1 loadings equal for all scale statistics?
3. Does PC1 follow $\sqrt{T}$ better than $T$?
4. Can a model classify the temperature with **PC1 only**?

```python
from sklearn.decomposition import PCA

pca = PCA().fit(X_clu)                                             # same 50-molecule samples, standardised
scores = pca.transform(X_clu)
clu["PC1"], clu["PC2"] = scores[:, 0], scores[:, 1]
clu["T"] = clu["Class"].map(dict(enumerate(TEMPERATURES)))

print("explained variance ratio:", np.round(pca.explained_variance_ratio_, 4))
loadings = pd.DataFrame(pca.components_[:2].T, index=FEATURES_B, columns=["PC1", "PC2"])
print(loadings.round(3), "\n")

pc1_by_T = clu.groupby("T")["PC1"].mean()
slope, intercept, r, p, se = stats.linregress(np.sqrt(pc1_by_T.index), pc1_by_T.values)
print("mean PC1 vs sqrt(T): r =", round(r, 5))
print("mean PC1 vs T      : r =", round(stats.pearsonr(pc1_by_T.index, pc1_by_T.values)[0], 5))
print("corr(PC1, sample mean speed) =", round(np.corrcoef(clu["PC1"], clu["mean"])[0, 1], 5))

pc_train, pc_test = scores[: len(samples[N_CLU][0])], scores[len(samples[N_CLU][0]):]
rf_pc1 = RandomForestClassifier(n_estimators=200, min_samples_leaf=20, random_state=SEED, n_jobs=-1)
rf_pc1.fit(pc_train[:, [0]], samples[N_CLU][0]["Class"])
print("RF test accuracy with PC1 only:", round(rf_pc1.score(pc_test[:, [0]], samples[N_CLU][1]["Class"]), 4))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for T in TEMPERATURES:
    part = clu[clu["T"] == T]
    axes[0].scatter(part["PC1"], part["PC2"], s=10, color=TEMP_COLORS[T], label=f"{T} K")
axes[0].set(xlabel=f"PC1 ({pca.explained_variance_ratio_[0]:.1%})",
            ylabel=f"PC2 ({pca.explained_variance_ratio_[1]:.1%})", title="PCA of 50-molecule samples")
axes[0].legend()
axes[1].plot(np.sqrt(pc1_by_T.index), pc1_by_T.values, "o", ms=8)
grid = np.linspace(10, 32, 50)
axes[1].plot(grid, intercept + slope * grid, color="black", label=f"linear fit, r = {r:.4f}")
axes[1].set(xlabel="sqrt(T) (K^0.5)", ylabel="mean PC1 score", title="PC1 follows sqrt(T)")
axes[1].legend()
plt.tight_layout()
plt.savefig("figures/pca_physics.png", dpi=200)
plt.show()
```

- `PCA().fit(X_clu)`: PCA on the same 800 standardised samples ($n = 50$) used for clustering. Like clustering, PCA does not use the labels.
- `pca.components_[:2].T`: the loadings of PC1 and PC2 for each feature.
- `stats.linregress(np.sqrt(T), mean PC1)`: a straight line between the average PC1 of each temperature and $\sqrt{T}$.
- `scores[: len(train)]`: the first rows are the train samples, the rest are the test samples (same order as `pd.concat`). The Random Forest uses only the PC1 column.
- **Result 1, one component is enough:** PC1 explains **96.5%** of the variance, PC2 only 2.1%, and the other six together about 1.4%. Eight statistics are reduced to **one number** with almost no loss.
- **Result 2, PC1 = the speed scale:** all 8 loadings of PC1 are almost equal (0.34 to 0.36). PC1 is simply "the general size of the speeds" in the sample. PC2 is a contrast between `q10` (+0.70) and `std` (−0.56), which describes the shape of the sample. Since the true shape does not depend on $T$, PC2 is mostly **random noise** (left plot: the colours separate along PC1, not along PC2).
- **Result 3, PC1 follows $\sqrt{T}$:** $r$ = **0.99957** with $\sqrt{T}$, against 0.99073 with $T$. PC1 also correlates 0.99929 with the sample mean speed. So PC1 behaves like $v_{mean} \propto \sqrt{T}$, the same law found by the regression in Cell 24.
- **Result 4, no loss of information:** a Random Forest with **PC1 only** has **99.38%** test accuracy, exactly the same as with all 8 features (Cell 27, $n = 50$) and equal to the Bayes ceiling.
- **Answer to RQ2:** yes, PCA preserves the physics. Without any physics input, it discovered that the Maxwell-Boltzmann distribution has **one degree of freedom** (the scale $a = \sqrt{k_BT/m}$), and its first component reproduces the $\sqrt{T}$ law. Figure saved as `figures/pca_physics.png`.

## Conclusion, limitations and references

### Answers to the research questions

**RQ1: Can machine learning classify the gas temperature from molecular speed distributions?**

- **One molecule (Level A): no, not reliably.** The Random Forest reached 49.6%, against a 25% baseline and a theoretical maximum (Bayes ceiling) of 50.5%. The limit comes from physics: the four Maxwell-Boltzmann curves overlap.
- **A sample of molecules (Level B): yes.** Accuracy grows with the sample size: 75% with 5 molecules, 95% with 20, 99.4% with 50 and 100% with 100. From $n = 5$ the model is equal to the Bayes ceiling, so it extracts all the information that exists in the data.
- Four different classifiers (Logistic Regression, kNN, Naive Bayes, Random Forest) give the same accuracy within 1 point. The bottleneck is the physics, not the algorithm.
- The most important feature, $\overline{v^2}$, is the physical sufficient statistic, linked to $T$ by equipartition ($\frac{1}{2}m\langle v^2\rangle = \frac{3}{2}k_BT$).
- Without labels, K-Means and Hierarchical clustering find the temperature groups with ARI = 0.96 to 0.97.

**RQ2: Does PCA preserve the physical structure?**

- **Yes.** PC1 explains 96.5% of the variance, all its loadings are equal, and it follows $\sqrt{T}$ ($r$ = 0.9996). A classifier using PC1 only keeps 99.4% accuracy. PCA discovered by itself that temperature acts through **one scale parameter**, $a = \sqrt{k_BT/m}$.

**Supporting results:**

- The simulated data matches theory: mean speeds within 2 standard errors, $\langle E \rangle \approx 1.5\,k_BT$, and Q-Q correlation ≥ 0.9998.
- Linear Regression learned $v_{mean} = 27.39\sqrt{T}$ (theory 27.49) and extrapolated to unseen hot gas with RMSE = 10.8 m/s, while the naive model in $T$ failed ($R^2 = -1.60$).
- The model predicted the **real** NIST speed of sound with a mean error of 0.32%.

### Limitations

- **Simulated data.** The speeds come from the ideal Maxwell-Boltzmann model, so the ML results show what is possible under ideal conditions. Real measurements would add instrument noise.
- **Ideal gas assumption.** At 100 K, N$_2$ is close to its boiling point (77 K), and the sound-speed error rises to 1.96%. Real-gas effects are not in the model.
- **Only four temperatures and one pressure (1 atm)** for classification. A finer temperature grid would make classification harder and closer to a regression problem.
- **One gas.** Other molecules have a different mass $m$, so the same speed corresponds to a different temperature. A model trained on N$_2$ would not transfer directly.
- **Statistics:** only Shapiro-Wilk and Q-Q plots were used to check distributions. A goodness-of-fit test (for example Kolmogorov-Smirnov) could be added in future work.

### Future work

- Use real molecular dynamics or experimental velocity data.
- Predict temperature as a **continuous** value (regression) from a sample of speeds.
- Train on several gases with mass as an extra feature.
